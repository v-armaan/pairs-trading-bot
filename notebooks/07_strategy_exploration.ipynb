```python
import alpaca
import pandas as pd
import sys
import os
sys.path.append(os.path.abspath(".."))
from src.data.alpaca_data import get_stock_bars
from src.data.preprocessing import prepare_price_data,split_data
from src.pairs.correlation import find_candidate_pairs
from src.pairs.selection import select_pairs
from src.pairs.spread import calculate_spread
from src.strategy.zscore import calculate_zscore
tickers = [
    "AAPL", "ADBE", "AMAT", "AMD", "AMZN",
    "AVGO", "CRM", "CSCO", "GOOGL", "IBM",
    "INTC", "META", "MSFT", "MU", "NFLX",
    "NOW", "NVDA", "ORCL", "QCOM", "TXN"
]
bars = get_stock_bars(tickers,"2021-01-01","2025-02-01")
prices_df=prepare_price_data(bars)
formation_prices,trading_prices=split_data(prices_df,"2023-12-29")
formation_returns=formation_prices.pct_change()
candidate_pairs=find_candidate_pairs(formation_returns,0.5)
selected_pairs=select_pairs(formation_prices,candidate_pairs)

trading_pairs=[]
for pair in selected_pairs:
    trading_spread=calculate_spread(trading_prices,(pair[0],pair[1],pair[2]),pair[4],pair[5])
    trading_pairs.append((*pair,trading_spread))
zscore_pairs=[]
for pair in trading_pairs:
    zscore=calculate_zscore(pair[6],calculate_spread(formation_prices,(pair[0],pair[1],pair[2]),pair[4],pair[5]))
    zscore_pairs.append((*pair,zscore))
    
#ZSCORE_PAIRS=[(STOCK1,STOCK2,COINTPVALUE,ADFPVALUE,ALPHA,BETA,TRADING_SPREAD,ZSCORES),(),(),...]



    
```
