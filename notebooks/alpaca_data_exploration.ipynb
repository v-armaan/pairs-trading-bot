```python
#LINEAR REGRESSION FOR BETA
import numpy as np

x = prices["MSFT"]
y = prices["AAPL"]

beta, alpha = np.polyfit(x, y, 1)#1 means degree 1 polynomialthat is a straight line,beta is the slope and alpha is the intercept

print("Hedge ratio:", beta)
print("Intercept:", alpha)
#this beta we use for hedge ratio
#Y=alpha+(beta)x+residual
#residual becomes our spread
```

```python
spread=prices["AAPL"]-(alpha+beta*prices["MSFT"])#residual is our spread
spread#how far is the actual apple price from the price the regression predicts
spread.plot(figsize=(12,6))#plot may have some mean reverting behaviour
```

correlation is just a prefilter real test for pair selection is cointegration and stationarity(adf test)

```python
#augmented-dickey fuller test on spread
from statsmodels.tsa.stattools import adfuller,coint
result=adfuller(spread)
print(f"ADF STATISTIC:{result[0]}")
print(f"p-value:{result[1]}")
```

p-value=0.279 we are keeping p<0.05 for pair selection therefore the test didnt give us enough evidence that the spread is stationary
1.correlation-told us that the pair has some relation
2.hedge ratio-gave us beta for our spread construction
3.adf to tell us if spread is statistically stationary


```python
#"Even though these two series individually wander around, do they share a long-run relationship such that some combination of them is stationary?"-cointegration
score,pvalue,_=coint(prices["AAPL"],prices["MSFT"])
print(f"coint statistics:{score}")
print(f"pvalue:{pvalue}")
```

```python
tickers = [
    "AAPL", "MSFT", "GOOGL", "AMZN", "META",
    "NVDA", "AVGO", "AMD", "NFLX", "ADBE",
    "INTC", "QCOM", "CSCO", "ORCL", "CRM",
    "IBM", "AMAT", "MU", "TXN", "NOW"
]
request = StockBarsRequest(
    symbol_or_symbols=tickers,
    timeframe=TimeFrame.Day,
    start="2021-01-01",
    end="2023-01-01"
)
bars=alpaca_client.get_stock_bars(request)
df=bars.df
df.index
```

```python
prices=df["close"].unstack(level="symbol")
prices
```

```python
returns = prices.pct_change()
returns
```

```python
corr_matrix=returns.corr()
corr_matrix
```

```python
#choosing correlation filter to be>=0.6(arbitrary,exact to be calculated later)
```

```python
candidate_pairs=[]
threshold=0.6
for i in range(len(corr_matrix.columns)):
    for j in range(i+1,len(corr_matrix.columns)):
        stock1=corr_matrix.columns[i]
        stock2=corr_matrix.columns[j]
        
        correlation=corr_matrix.iloc[i,j]
        if correlation>=threshold:
            candidate_pairs.append((stock1,stock2,correlation))
print(len(candidate_pairs))
print(candidate_pairs)      
```

```python
coint_pairs=[]
for pair in candidate_pairs:
    cstat,pvalue,_=coint(prices[pair[0]],prices[pair[1]])
    if pvalue<0.05:
        coint_pairs.append((pair[0],pair[1],pvalue))
coint_pairs
```

```python
#checking mean reverting behaviour of the spread 
#step1)make spreads for each pair
#step2)check pvalue from adf for each pair
#step3)only keep those with p<0.05
adf_passed=[]
for pair in coint_pairs:
    stock1=pair[0]
    stock2=pair[1]
    beta,alpha=np.polyfit(prices[stock1],prices[stock2],1)
    spread=prices[stock2]-(alpha+beta*prices[stock1])
    adresults=adfuller(spread)
    print(adresults[1])
    if adresults[1]<0.05:
        adf_passed.append((stock2,stock1,adresults[1],beta,alpha))
adf_passed
    
```

```python
#AMAT AND AMD PASSED BOTH THE TESTS NOW TO CHECK HALF LIFE OF MEAN REVERSION
#FOR HALF LIFE
#WE WANT ΔSt​=St​−St−1​ WHICH IS the first different of my spread series
#then we want a relation between todays change and yesterdays spread ΔSt​=λSt−1​+ϵt​
#so X=spread shifted by 1,Y=change in spread
#then run a simple regression Y=(LAMBDA)X+EPSILON
```

```python
spread_selectedpairs=[]
for pair in adf_passed:
    spread=prices[pair[0]]-(pair[4]+pair[3]*prices[pair[1]])
    spread_selectedpairs.append((pair[0],pair[1],pair[2],pair[3],spread))
spread_selectedpairs
```

```python

import math
hl_pairs=[]
for pair in spread_selectedpairs:
    x=pair[4].shift(1)
    y=pair[4].diff()
    combined=pd.concat([x,y],axis=1)
    combined.dropna(inplace=True)
    slope,intercept=np.polyfit(combined.iloc[:,0],combined.iloc[:,1],1)
    print("slope=",slope,"\n","intercept=",intercept)
    half_life=abs(math.log(2,math.e)/slope)
    hl_pairs.append((pair[0],pair[1],pair[2],pair[3],pair[4],half_life))
hl_pairs
```

```python
#half life is 34.87 which is w6.29 trading days
#for this project 5<=half life<=90
```

```python
#ZSCORE-HOW MANY STANDARD DEVIATIONS IS A PARTICUALR VALUE AWAY FROM THE MEAN
#basically if z=+2 then current spread is 2 std above mean,if -2 then current spread is 2 std below mean
#Z=(SPREADT-mean(spread))/std(Spread)
```

```python
z=(spread-spread.mean())/spread.std()
z
#!!!!!LOOKAHEAD BIAS!!!!!
```

```python
#entry threshold-at what z score do i want to bet on mean reversion
#exit-at what z score do i want to close my positions
#long-buy first sell later
#short-sell first buy later
```

```python
#say if Z>+2,the spread is unusually +ve,short Y long betaX
#say if Z<-2,the spread is unusually -ve,short betaX long Y
#exit at z->0
#NEED A STOP LOSS(IF THE MEAN DOESNT REVERT AND |Z| KEEPS ON INCREASING)-exit at this too
```

```python
#CALCULATING Z FOR PAIRS WITHOUT LOOKAHEAD BIAS(i.e calc z score for a day using information available before that day only)
count=1
for pair in hl_pairs:
    print(f"Spread {count}:{pair[4]}")
    count+=1
```

```python
#first calculating for the first pair only
sample=hl_pairs[0][4].copy()
sample
```

```python
#ZSCORE WITHOUT LOOK AHEAD BIAS
zscores=[]
counter=0
for i in sample:
    previous=sample.iloc[:counter]
    today=sample.iloc[counter]
    zscores.append((today-previous.mean())/previous.std())
    counter+=1
print(zscores)
```

```python
#we get a z score of 7 on the 4th day,we should avoid this becuase this happens due to very little data at that point for calculation of z score so avoid trading on early z scores
```

```python
z_pairs=[]
for pair in hl_pairs:
    zscores=[]
    counter=0
    sample=pair[4]
    for i in sample:
        previous=sample.iloc[:counter]
        today=sample.iloc[counter]
        zscores.append((today-previous.mean())/previous.std())
        counter+=1
    z_pairs.append((pair[0],pair[1],pair[2],pair[3],pair[4],pair[5],pd.Series(zscores,index=sample.index)))
z_pairs
```

```python
# #the warmup period for trading using z score is tied to the half life that is start trading after about 3 half lives
# 1 half-life - one typical mean-reversion timescale
# 2 half-lives - more history
# 3 half-lives -enough time to observe several
#                mean-reversion cycles
#warmup is a trading rule filter not a z score calc rule
```
