```python
from src.pairs.selection import select_pairs
formation_returns=formation_data.pct_change().dropna()
corr_matrix=formation_returns.corr()
candidate_pairs=[]
correlation_threshold=0.6
for i in range(len(corr_matrix.columns)):
    for j in range(i+1,len(corr_matrix.columns)):
        stock1=corr_matrix.columns[i]
        stock2=corr_matrix.columns[j]
        correlation=corr_matrix.iloc[i,j]
        if correlation>=correlation_threshold:
            candidate_pairs.append(
                (stock1,stock2,correlation)
            )
print(len(candidate_pairs))
selected_pairs=select_pairs(formation_data,candidate_pairs)
selected_pairs
```

```python
from src.pairs.spread import calculate_spread

formation_spreads = []

for pair in selected_pairs:

    spread = calculate_spread(
        formation_data,
        pair,
        pair[4],   # alpha
        pair[5]    # beta
    )

    formation_spreads.append(
        (pair[0], pair[1], pair[2], pair[3], pair[4], pair[5], spread)
    )
formation_spreads
```

```python
#calculation z scor for formation period
from src.strategy.zscore import calculate_zscore
from src.strategy.signals import calculate_thresholds

formation_zscores = []

for pair in formation_spreads:

    zscore = calculate_zscore(pair[6])

    lower_threshold, upper_threshold = calculate_thresholds(zscore)

    formation_zscores.append(
        (
            pair[0],
            pair[1],
            pair[2],
            pair[3],
            pair[4],
            pair[5],
            pair[6],
            zscore,
            lower_threshold,
            upper_threshold
        )
    )
formation_zscores
```

```python
position = 0

position_df = pd.DataFrame(trading_zscore.copy())
position_df = position_df.rename(columns={0: "zscore"})

position_df["position"] = 0
position_df["event"] = "NO_EVENT"

entry_day = None

# Formation-period thresholds
lower_threshold = formation_zscores[0][8]
upper_threshold = formation_zscores[0][9]
formation_median = formation_zscores[0][7].quantile(0.5)

for i in range(len(position_df)):

    current_zscore = position_df.iloc[i, 0]

    # -------------------------
    # FLAT: look for new entry
    # -------------------------
    if position == 0:

        if current_zscore < lower_threshold:
            position = 1
            entry_day = i

            position_df.loc[position_df.index[i], "event"] = "LONG_ENTRY"

        elif current_zscore > upper_threshold:
            position = -1
            entry_day = i

            position_df.loc[position_df.index[i], "event"] = "SHORT_ENTRY"


    # -------------------------
    # LONG POSITION
    # -------------------------
    elif position == 1:

        # Mean reversion
        if current_zscore >= formation_median:
            position = 0
            entry_day = None

            position_df.loc[
                position_df.index[i], "event"
            ] = "LONG_MEAN_EXIT"

        # Stop loss
        elif current_zscore <= lower_threshold - 1:
            position = 2
            entry_day = None

            position_df.loc[
                position_df.index[i], "event"
            ] = "LONG_STOPLOSS"

        # Maximum holding period
        elif i - entry_day >= max_holding_period:
            position = 0
            entry_day = None

            position_df.loc[
                position_df.index[i], "event"
            ] = "LONG_MAX_HOLDING"


    # -------------------------
    # SHORT POSITION
    # -------------------------
    elif position == -1:

        # Mean reversion
        if current_zscore <= formation_median:
            position = 0
            entry_day = None

            position_df.loc[
                position_df.index[i], "event"
            ] = "SHORT_MEAN_EXIT"

        # Stop loss
        elif current_zscore >= upper_threshold + 1:
            position = -2
            entry_day = None

            position_df.loc[
                position_df.index[i], "event"
            ] = "SHORT_STOPLOSS"

        # Maximum holding period
        elif i - entry_day >= max_holding_period:
            position = 0
            entry_day = None

            position_df.loc[
                position_df.index[i], "event"
            ] = "SHORT_MAX_HOLDING"


    # -------------------------
    # COOLDOWN AFTER LONG STOP
    # -------------------------
    elif position == 2:

        # Wait until z-score comes back
        # inside the long-entry region
        if current_zscore > lower_threshold:
            position = 0

            position_df.loc[
                position_df.index[i], "event"
            ] = "LONG_REENTRY_UNLOCK"


    # -------------------------
    # COOLDOWN AFTER SHORT STOP
    # -------------------------
    elif position == -2:

        # Wait until z-score comes back
        # inside the short-entry region
        if current_zscore < upper_threshold:
            position = 0

            position_df.loc[
                position_df.index[i], "event"
            ] = "SHORT_REENTRY_UNLOCK"


    # Store current state
    position_df.loc[
        position_df.index[i], "position"
    ] = position


print(
    f"Upper threshold: {upper_threshold}\n"
    f"Lower threshold: {lower_threshold}\n"
    f"Formation median: {formation_median}"
)

pd.set_option("display.max_rows", None)

position_df
```
